```python
from chatterbot import ChatBot
from chatterbot.trainers import ListTrainer

# Initialize chatbot with custom settings
wellness_bot = ChatBot(
    'WellnessBot',
    read_only=True,
    logic_adapters=['chatterbot.logic.BestMatch']
)

# Create a trainer for the chatbot
trainer = ListTrainer(wellness_bot)

# Define a list of health-related conversation pairs
dialogues = [
    'Hi', 'Hello! I am WellnessBot. How can I help you today?',
    'Hello', 'Hi there! Do you have any health-related concerns?',
    'I have a headache', 'Try to rest, drink water, and take a mild painkiller if necessary.',
    'What should I do if I have a fever?', 'Keep yourself hydrated and rest well. If symptoms persist, see a healthcare provider.',
    'I feel dizzy', 'Sit or lie down, take deep breaths, and drink water. Seek help if it doesn’t improve.',
    'What should I eat for a cold?', 'Consume warm fluids, soups, and fruits rich in vitamin C.',
    'How to stay healthy?', 'Maintain a balanced diet, stay active, drink enough water, and get quality sleep.',
    'What to do in case of a cut?', 'Clean it thoroughly, apply antiseptic, and cover it with a clean bandage.',
    'How much water is enough daily?', 'Aim for 2–3 liters a day, but individual needs may vary.',
    'Thank you', 'You’re most welcome! Wishing you good health.',
    'Bye', 'Take care! Feel free to ask anytime.'
]

# Train the chatbot with the predefined dialogues
trainer.train(dialogues)

# Start the chatbot interaction
print("Welcome! You can ask me health questions. Type 'exit' to quit.")

while True:
    query = input("You: ")
    if query.lower() == 'exit':
        print("WellnessBot: Take care! See you next time.")
        break
    reply = wellness_bot.get_response(query)
    print(f"WellnessBot: {reply}")
```
